```python
import random

# Part A - Machine Initialization
machines = ['A', 'B', 'C']

# Number of plays
N_play = 10

# Exploration probability
epsilon = 0.2

# Fixed seed for reproducible result
random.seed(38)

# Part B - Initialize N, R and Q
N = [0, 0, 0]
R = [0, 0, 0]
Q = [0.0, 0.0, 0.0]

# Given reward patterns
reward_patterns = {
    'A': [3, 4, 2, 5],
    'B': [6, 7, 5, 8],
    'C': [4, 3, 5, 4]
}

print("EPSILON-GREEDY MULTI-ARMED BANDIT")
print("-----------------------------------")
print("Epsilon =", epsilon)

# Perform 10 plays
for play in range(N_play):

    # Generate random number
    r = random.random()

    # Epsilon-Greedy Decision
    if r < epsilon:

        # Exploration
        machine_index = random.randint(0, 2)
        decision = "Exploration"

    else:

        # Exploitation
        machine_index = Q.index(max(Q))
        decision = "Exploitation"

    machine = machines[machine_index]

    # Get reward according to given reward pattern
    reward_number = N[machine_index]
    reward = reward_patterns[machine][reward_number]

    # Update participation count
    N[machine_index] += 1

    # Store reward
    R[machine_index] = reward

    # Incremental sample-average update
    Q[machine_index] = (
        Q[machine_index]
        + (1 / N[machine_index])
        * (reward - Q[machine_index])
    )

    print("\nPlay", play + 1)
    print("Random number =", round(r, 4))
    print("Decision =", decision)
    print("Selected Machine =", machine)
    print("Reward =", reward)

    print("N =", N)
    print("R =", R)
    print("Q =", [round(x, 2) for x in Q])


# Final Results
print("\n===================================")
print("FINAL RESULTS")
print("===================================")

print("Machine\tN\tR\tQ")

for i in range(3):
    print(
        machines[i],
        "\t",
        N[i],
        "\t",
        R[i],
        "\t",
        round(Q[i], 2)
    )

# Find best machine
best_index = Q.index(max(Q))

print("\nBest Machine =", machines[best_index])
print("Highest Q-value =", round(Q[best_index], 2))
```
