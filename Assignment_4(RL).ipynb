```python
import math

# Part A – Machine Initialization
machines = ['A', 'B', 'C']

N_play = 10

# Part B – Initialize N, R, UCB and Q tables
N = {machine: 0 for machine in machines}       # Number of times played
R = {machine: 0 for machine in machines}       # Total reward
UCB = {machine: 0 for machine in machines}     # UCB values
Q = {machine: 0.0 for machine in machines}     # Estimated reward

print("=== Upper Confidence Bound (UCB) ===")
print("Machines:", machines)
print("Number of plays:", N_play)

# Part C, D and E – Play machines, collect rewards and update Q
for play in range(1, N_play + 1):

    # Calculate UCB
    for machine in machines:
        if N[machine] == 0:
            UCB[machine] = float('inf')
        else:
            UCB[machine] = Q[machine] + math.sqrt(
                (2 * math.log(play)) / N[machine]
            )

    # Select machine
    # First play each machine once
    if play <= len(machines):
        selected_machine = machines[play - 1]
    else:
        selected_machine = max(UCB, key=UCB.get)

    print("\nPlay", play)
    print("Selected Machine:", selected_machine)

    # Part D – Get reward from user
    reward = float(input("Enter reward: "))

    # Part E – Update N, R and Q
    N[selected_machine] += 1
    R[selected_machine] += reward

    # Incremental sample-average formula:
    # Q_new = Q_old + (Reward - Q_old) / N
    Q[selected_machine] = Q[selected_machine] + (
        reward - Q[selected_machine]
    ) / N[selected_machine]

# Calculate final UCB values
for machine in machines:
    if N[machine] == 0:
        UCB[machine] = float('inf')
    else:
        UCB[machine] = Q[machine] + math.sqrt(
            (2 * math.log(N_play)) / N[machine]
        )

# Part F – Display Results
print("\n========== FINAL RESULTS ==========")

print("\nMachine\tN (Count)\tR (Reward)\tUCB\t\tQ-Value")

for machine in machines:
    print(
        f"{machine}\t"
        f"{N[machine]}\t\t"
        f"{R[machine]:.2f}\t\t"
        f"{UCB[machine]:.4f}\t\t"
        f"{Q[machine]:.4f}"
    )

# Total reward received
total_reward = sum(R.values())

print("\nTotal Reward Received:", total_reward)

# Best machine
best_machine = max(Q, key=Q.get)
print("Best Machine based on Q-value:", best_machine)
```
