```python
import gymnasium as gym
import pandas as pd
```

```python
def random_agent(env_name):
    # 1. Create the environment
    env = gym.make(env_name)

    # 2. Reset the environment
    state, info = env.reset()

    total_reward = 0
    steps = 0

    print("Environment:", env_name)
    print("Initial State:", state)
    print("Action Space:", env.action_space)
    print("-" * 50)

    # Continue until episode terminates
    terminated = False
    truncated = False

    while not (terminated or truncated):

        # Select a random action
        action = env.action_space.sample()

        # Execute action
        next_state, reward, terminated, truncated, info = env.step(action)

        # Display interaction
        print(
            f"Step: {steps + 1} | "
            f"Action: {action} | "
            f"Reward: {reward} | "
            f"Next State: {next_state}"
        )

        # Update values
        total_reward += reward
        steps += 1
        state = next_state

    # Calculate average reward
    average_reward = total_reward / steps

    print("-" * 50)
    print("Episode Finished")
    print("Total Steps:", steps)
    print("Total Reward:", total_reward)
    print("Average Reward:", average_reward)

    env.close()

    return total_reward, average_reward
```

```python
cartpole_reward, cartpole_average = random_agent("CartPole-v1")
```

```python
mountaincar_reward, mountaincar_average = random_agent("MountainCar-v0")
```

```python
frozenlake_reward, frozenlake_average = random_agent("FrozenLake-v1")
```

```python
print("\n========== FINAL RESULTS ==========")

print("CartPole")
print("Total Reward:", cartpole_reward)
print("Average Reward:", cartpole_average)

print("\nMountainCar")
print("Total Reward:", mountaincar_reward)
print("Average Reward:", mountaincar_average)

print("\nFrozenLake")
print("Total Reward:", frozenlake_reward)
print("Average Reward:", frozenlake_average)
```
