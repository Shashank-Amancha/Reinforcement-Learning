```python

# Assignment 1: Bellman Expectation Equations
# Stochastic policy, deterministic environment

gamma = 0.9

# Bellman equations:
# V0 = -1 + 0.81 * V1
# V1 = 89.9 + 0.09 * V0

# Solve algebraically
V0 = 71.819 / (1 - 0.0729)
V1 = 89.9 + 0.09 * V0

# Calculate action values
Q0_down = -1 + gamma * V1
Q0_right = -1 + gamma * 0

Q1_right = 100 + gamma * 0
Q1_up = -1 + gamma * V0

# Verify state values using policy probabilities
V0_check = 0.9 * Q0_down + 0.1 * Q0_right
V1_check = 0.9 * Q1_right + 0.1 * Q1_up

print("Assignment 1 Results")
print("--------------------")
print(f"V(S0) = {V0:.2f}")
print(f"V(S1) = {V1:.2f}")
print(f"Q(S0, Down) = {Q0_down:.2f}")
print(f"Q(S0, Right) = {Q0_right:.2f}")
print(f"Q(S1, Right) = {Q1_right:.2f}")
print(f"Q(S1, Up) = {Q1_up:.2f}")

print("\nVerification")
print(f"V(S0) check = {V0_check:.2f}")
print(f"V(S1) check = {V1_check:.2f}")
```

```python

# Assignment 2: Iterative Policy Evaluation

gamma = 0.9
theta = 0.000001

# Initialize state values
V0 = 0.0
V1 = 0.0

iterations = 0

# Iterative policy evaluation
while True:
    # Use old values to calculate both new values
    new_V0 = (
        0.1 * (-1 + gamma * 0)
        + 0.9 * (-1 + gamma * V1)
    )

    new_V1 = (
        0.9 * (100 + gamma * 0)
        + 0.1 * (-1 + gamma * V0)
    )

    # Calculate maximum change
    delta = max(
        abs(new_V0 - V0),
        abs(new_V1 - V1)
    )

    # Update state values together
    V0 = new_V0
    V1 = new_V1

    iterations += 1

    # Stop when convergence is reached
    if delta < theta:
        break

# Calculate action values after convergence

# S0, Down: Wall (0.1), S1 (0.9)
Q_S0_Down = (
    0.1 * (-1 + gamma * 0)
    + 0.9 * (-1 + gamma * V1)
)

# S0, Right: Wall (1.0)
Q_S0_Right = -1 + gamma * 0

# S1, Right: Goal (0.9), S0 (0.1)
Q_S1_Right = (
    0.9 * (100 + gamma * 0)
    + 0.1 * (-1 + gamma * V0)
)

# S1, Up: S0 (1.0)
Q_S1_Up = -1 + gamma * V0

# Display results
print("Assignment 2: Results")
print("Iterations:", iterations)
print(f"V(S0) = {V0:.8f}")
print(f"V(S1) = {V1:.8f}")

print(f"Q(S0, Down) = {Q_S0_Down:.8f}")
print(f"Q(S0, Right) = {Q_S0_Right:.8f}")
print(f"Q(S1, Right) = {Q_S1_Right:.8f}")
print(f"Q(S1, Up) = {Q_S1_Up:.8f}")

# Verify policy consistency
print("\nPolicy Consistency Checks")

print(
    "V(S0) = Q(S0, Down):",
    abs(V0 - Q_S0_Down) < theta
)

print(
    "V(S1) = Q(S1, Right):",
    abs(V1 - Q_S1_Right) < theta
)
```
