```python
import random

# Part A: Machine Initialization
machines = ['A', 'B', 'C']

# Given reward patterns
reward_patterns = {
    'A': [3, 4, 2, 5],
    'B': [6, 7, 5, 8],
    'C': [4, 3, 5, 4]
}

# Part B: Initialize N, R and Q tables
N = {machine: 0 for machine in machines}
R = {machine: 0.0 for machine in machines}
Q = {machine: 0.0 for machine in machines}

epsilon = 0.2
number_of_plays = 10
total_reward = 0.0

print("Modified Epsilon-Greedy Multi-Armed Bandit")
print("Number of plays:", number_of_plays)
print("Epsilon:", epsilon)

print("\nReward patterns:")
print("A: 3, 4, 2, 5")
print("B: 6, 7, 5, 8")
print("C: 4, 3, 5, 4")

# Keep track of which reward to use next for each machine
reward_index = {
    'A': 0,
    'B': 0,
    'C': 0
}

# Part C: ε-Greedy Decision
for play in range(1, number_of_plays + 1):

    # Find highest Q-value
    highest_q = max(Q.values())

    # Machines having the highest Q-value
    best_machines = [
        machine for machine in machines
        if Q[machine] == highest_q
    ]

    # Randomly choose one if there is a tie
    best_machine = random.choice(best_machines)

    # Generate random number
    r = random.random()

    if r < epsilon:
        # Exploration
        other_machines = [
            machine for machine in machines
            if machine != best_machine
        ]

        selected_machine = random.choice(other_machines)
        decision = "Exploration"

    else:
        # Exploitation
        selected_machine = best_machine
        decision = "Exploitation"

    print(f"\nPlay {play}: {decision}")
    print("Selected machine:", selected_machine)

    # Part D: Automatically get reward
    index = reward_index[selected_machine]

    if index < len(reward_patterns[selected_machine]):
        reward = reward_patterns[selected_machine][index]
        reward_index[selected_machine] += 1
    else:
        # If a machine is selected more than 4 times
        print("No more predefined rewards for this machine.")
        reward = 0

    print("Reward received:", reward)

    # Part E: Incremental sample-average Q update
    N[selected_machine] += 1
    R[selected_machine] += reward

    Q[selected_machine] = Q[selected_machine] + (
        reward - Q[selected_machine]
    ) / N[selected_machine]

    total_reward += reward

# Part F: Display final results
print("\n--- Final N, R, Q Tables ---")
print("Machine\tN (plays)\tR (total reward)\tQ (average reward)")

for machine in machines:
    print(
        f"{machine}\t"
        f"{N[machine]}\t\t"
        f"{R[machine]:.2f}\t\t\t"
        f"{Q[machine]:.2f}"
    )

print("\nTotal reward received by the player:", total_reward)
```
