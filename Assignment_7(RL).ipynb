# **Assignment-1**

```python
n = int(input("Enter number of time steps: "))

rewards = []

for i in range(n):
    r = float(input("Enter reward: "))
    rewards.append(r)

gamma = float(input("Enter discount factor gamma: "))

undiscounted_return = sum(rewards)

discounted_return = 0

for i in range(n):
    discounted_return = discounted_return + (gamma ** i) * rewards[i]

print("\nEntered reward values:", rewards)
print("Undiscounted return:", undiscounted_return)
print("Discounted return:", discounted_return)
```

```python
%%writefile grid_env_det.py

states = ["S0", "S1", "S3", "S4", "G"]

actions = ["Up", "Down", "Left", "Right"]

reward = {
    "normal": -1,
    "goal": 10,
    "obstacle": -5
}

gamma = 0.9

transition = {
    "S0": {
        "Up": "S0",
        "Down": "S0",
        "Left": "S0",
        "Right": "S1"
    },

    "S1": {
        "Up": "S1",
        "Down": "S3",
        "Left": "S0",
        "Right": "G"
    },

    "S3": {
        "Up": "S1",
        "Down": "S3",
        "Left": "S3",
        "Right": "S4"
    },

    "S4": {
        "Up": "G",
        "Down": "S4",
        "Left": "S3",
        "Right": "S4"
    }
}


def get_next_state(current_state, action):
    return transition[current_state][action]
```
