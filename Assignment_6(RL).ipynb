```python
import gymnasium as gym
import random

# Create CliffWalking environment
env = gym.make("CliffWalking-v1")

# Number of episodes
episodes = 10

total_rewards = []

for episode in range(1, episodes + 1):

    # Reset environment
    state, info = env.reset()

    total_reward = 0
    terminated = False
    truncated = False

    while not (terminated or truncated):

        # Get number of available actions
        number_of_actions = env.action_space.n

        # Select a random action
        action = random.randrange(number_of_actions)

        # Perform the action
        next_state, reward, terminated, truncated, info = env.step(action)

        # Add reward
        total_reward += reward

        # Move to next state
        state = next_state

    # Store episode reward
    total_rewards.append(total_reward)

    print("Episode", episode, "Total Reward:", total_reward)

# Calculate average reward
average_reward = sum(total_rewards) / len(total_rewards)

print("\nAverage Reward:", average_reward)

env.close()
```

```python
import gymnasium as gym
import random

# Create Taxi environment
env = gym.make("Taxi-v4")

# Number of episodes
episodes = 10

total_rewards = []

for episode in range(1, episodes + 1):

    # Reset environment
    state, info = env.reset()

    total_reward = 0
    terminated = False
    truncated = False

    while not (terminated or truncated):

        # Identify available actions
        number_of_actions = env.action_space.n

        # Select a random action
        action = random.randrange(number_of_actions)

        # Execute action
        next_state, reward, terminated, truncated, info = env.step(action)

        # Add reward
        total_reward += reward

        # Update state
        state = next_state

    # Store total reward
    total_rewards.append(total_reward)

    # Print reward for each episode
    print("Episode", episode, "Total Reward:", total_reward)

# Calculate average reward
average_reward = sum(total_rewards) / len(total_rewards)

print("\nAverage Reward:", average_reward)

env.close()
```
