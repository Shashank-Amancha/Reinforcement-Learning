```python
%%writefile grid_env_det.py

states = ["S0", "S1", "S3", "S4", "G"]

actions = ["Up", "Down", "Left", "Right"]

reward = {
    "normal": -1,
    "goal": 10,
    "obstacle": -5
}

gamma = 0.9

transition = {
    "S0": {
        "Up": "S0",
        "Down": "S0",
        "Left": "S0",
        "Right": "S1"
    },

    "S1": {
        "Up": "S1",
        "Down": "S3",
        "Left": "S0",
        "Right": "G"
    },

    "S3": {
        "Up": "S1",
        "Down": "S3",
        "Left": "S3",
        "Right": "S4"
    },

    "S4": {
        "Up": "G",
        "Down": "S4",
        "Left": "S3",
        "Right": "S4"
    }
}


def get_next_state(current_state, action):
    return transition[current_state][action]
```

```python
import grid_env_det
```

```python
policy = {
    "S0": "Right",
    "S1": "Down",
    "S3": "Right",
    "S4": "Up"
}

print("Policy:", policy)
```

```python
def state_value_function(state):

    total = 0
    discount = 1

    while state != "G":

        action = policy[state]
        next_state = grid_env_det.get_next_state(state, action)

        if next_state == "G":
            reward = 10
        else:
            reward = -1

        total = total + discount * reward
        discount = discount * grid_env_det.gamma
        state = next_state

    return total


def action_value_function(state, action):

    total = 0
    discount = 1

    while True:

        next_state = grid_env_det.get_next_state(state, action)

        if next_state == "G":
            reward = 10
        else:
            reward = -1

        total = total + discount * reward

        if next_state == "G":
            break

        discount = discount * grid_env_det.gamma
        state = next_state
        action = policy[state]

    return total


print("State Value for S0:", state_value_function("S0"))
print("Action Value for S0 and Up:", action_value_function("S0", "Up"))
```
